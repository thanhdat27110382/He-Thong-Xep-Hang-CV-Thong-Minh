```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd
```

```python
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")
options.add_argument("--disable-blink-features=AutomationControlled")
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)
```

```python
url = "https://resumeworded.com/resume-examples#experienced-hires"
driver.get(url)
```

```python
def scroll_to_bottom():
    last_height = driver.execute_script("return document.body.scrollHeight")
    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == last_height:
            break
        last_height = new_height

scroll_to_bottom()
```

```python
data = []
```

```python
containers = driver.find_elements(By.CSS_SELECTOR, "div.right-small-jd-link-container")
print(f"Tìm thấy {len(containers)} khối right-small-jd-link-container")
```

```python
for container in containers:
    try:
        list_items = container.find_elements(By.TAG_NAME, "li")
        for li in list_items:
            try:
                a_tag = li.find_element(By.TAG_NAME, "a")
                href = a_tag.get_attribute("href")

                try:
                    category_div = a_tag.find_element(By.CLASS_NAME, "single-jd-link-container")
                    category = category_div.text.strip()
                except:
                    category = ""

                if href and category:
                    data.append({
                        "Links": href,
                        "Category": category
                    })
            except Exception as e:
                print(f"Lỗi khi xử lý <li>: {e}")
    except Exception as e:
        print(f"Lỗi khi xử lý container: {e}")
```

```python
driver.quit()
```
