```python
import pandas as pd
import re

# Hàm kiểm tra ký tự bất hợp pháp (Excel không cho phép)
illegal_pattern = re.compile(r'[\x00-\x08\x0B\x0C\x0E-\x1F]')

def has_illegal_chars(value):
    if isinstance(value, str):
        return bool(illegal_pattern.search(value))
    return False

# Giả sử df_mer là DataFrame của bạn
df = df_mer.copy()

# Danh sách ô bị lỗi
illegal_cells = []

for row_idx, row in df.iterrows():
    for col in df.columns:
        val = row[col]
        if has_illegal_chars(val):
            illegal_cells.append((row_idx, col, repr(val)))  # repr để hiện ký tự ẩn

# In kết quả
if illegal_cells:
    print(f"🔴 Phát hiện {len(illegal_cells)} ô có ký tự không hợp lệ:")
    for i, (r, c, v) in enumerate(illegal_cells[:10]):  # chỉ in 10 lỗi đầu
        print(f"  Hàng {r}, Cột '{c}': {v}")
    if len(illegal_cells) > 10:
        print(f"... (và {len(illegal_cells) - 10} ô khác)")
else:
    print("✅ Không phát hiện ký tự bất hợp lệ nào.")
```

```python
# 🧹 Tuỳ chọn: làm sạch toàn bộ ký tự lỗi
def clean_illegal_chars(value):
    if isinstance(value, str):
        return illegal_pattern.sub('', value)
    return value

df_clean = df.applymap(clean_illegal_chars)
```

```python
import pandas as pd
import re

input_path = "D:/Downloads/train-00000-of-00001.parquet"
output_path = "D:/Downloads/data_new_6.xlsx"

# Đọc parquet
df = pd.read_parquet(input_path)

# Regex loại bỏ ký tự điều khiển không hợp lệ
illegal_pattern = re.compile(r'[\x00-\x08\x0B\x0C\x0E-\x1F]')

def clean_illegal_chars(value):
    if isinstance(value, str):
        return illegal_pattern.sub('', value)
    return value

# Áp dụng cho toàn bộ DataFrame
df_clean = df.applymap(clean_illegal_chars)

# Ghi ra Excel sau khi làm sạch
df_clean.to_excel(output_path, index=False, engine='openpyxl')

print(f"✅ Đã làm sạch và lưu file: {output_path}")
```
